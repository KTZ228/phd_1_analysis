```python
import numpy as np
import pandas as pd
import os
import pickle as pkl
```

```python
file_list = os.listdir(os.getcwd())
```

```python
save_df=pd.read_csv('sess1_allsubs_sumdata_rev_phase.csv')
```

```python
def get_order(sub_df):
    first_stim = sub_df["happ_vol_stab"].iloc[15]
    if first_stim == 1:
        return 'vol_first'
    elif first_stim == 0:
        return 'sta_first'
    else:
        return "unknown"
```

```python
select_df = select_df[select_df['subjectively_correct']!='Late']
select_df['unit'] = select_df['sub_num'].astype(str) + '_ses-' + select_df['session'].astype(str)
select_df['order'] = select_df.groupby('sub_num').apply(get_order)
#select_df = select_df[select_df['order']=='sta_first']
```

```python
select_df = select_df[select_df['objectively_correct']!='Even']
```

```python
codes, uniques = pd.factorize(select_df['unit'], sort=False)
```

```python
save_fit = {
    'ns':len(uniques),
    'nt':len(codes),
    'sub_list':codes.astype('int')+1,
    'choice':select_df['response'].map({'up':0,'down':1}).astype('int'),
    'vol':((select_df['happ_vol_stab']+select_df['ang_vol_stab'])>.1)*1- .5,
    'cue':select_df['valence'].map({'Angry':-.5,'Happy':.5}),
    'outcome':select_df['subjectively_correct'].map({'False':0,'True':1}).astype('int')
    
}
```

```python
with open('included_session1_stan_fit.pkl','wb') as f:
    pkl.dump(save_fit,f)
```
